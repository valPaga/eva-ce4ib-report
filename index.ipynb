## Building

Single-storey office (SIA 2024 room type 3.01) above an unheated basement, 30 × 15 m, split into a south zone and a
north zone of 225 m² each. External walls 0.20 m mineral wool + 0.20 m concrete, roof 0.16 m PUR + 0.28 m concrete,
floor 0.14 m mineral wool + 0.30 m concrete, triple glazing on 60 % of each façade with external blinds.
Weather: Chur, TMYx 2004–2018.
